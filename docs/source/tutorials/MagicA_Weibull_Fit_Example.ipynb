# MagicA: Weibull Fit and Goodness-of-Fit Evaluation

This notebook demonstrates how to use MagicA to fit a Weibull distribution to wind speed data and evaluate the fit using three statistical methods: Chi-Squared, Kolmogorov-Smirnov, and Root Mean Square. It also provides API documentation for the main MagicA classes.

```python
# Import MagicA and required libraries
import magica as ma
import numpy as np
```

```python
# Prepare example wind speed data
wind_data = [2.1, 5.4, 8.7, 12.3, 6.8, 9.1, 15.2, 3.4, 7.6, 11.0, 4.5, 13.2, 8.9, 6.7, 10.5]
```

```python
# Load data with MagicA
ma_data = ma.read_data(wind_data)
print(ma_data)
```

```python
# Get basic statistics
stats = ma_data.get_basic_stats()
print('Basic statistics:')
for k, v in stats.items():
    print(f'  {k}: {v}')
```

```python
# Fit Weibull distribution
fitted = ma_data.fit_distribution('weibull')
print(fitted)
```

```python
# Get fitted Weibull parameters
params = fitted.get_fitted_params()
print('Weibull parameters:', params)
```

```python
# Get distribution information
info = fitted.get_distribution_info()
print('Distribution information:')
for k, v in info.items():
    print(f'  {k}: {v}')
```

```python
# Evaluate fit: Chi-Squared test
chi2 = fitted.goodness_of_fit('chi2')
print('Chi-Squared test:')
for k, v in chi2.items():
    print(f'  {k}: {v}')
```

```python
# Evaluate fit: Kolmogorov-Smirnov test
ks = fitted.goodness_of_fit('ks')
print('Kolmogorov-Smirnov test:')
for k, v in ks.items():
    print(f'  {k}: {v}')
```

```python
# Evaluate fit: Root Mean Square test
rms = fitted.goodness_of_fit('rms')
print('Root Mean Square test:')
for k, v in rms.items():
    print(f'  {k}: {v}')
```
